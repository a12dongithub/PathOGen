# PathOGen conditional inference smoke test

Recommended runtime: **L4 GPU**. Choose **Runtime → Change runtime type → GPU**, then run all cells. GPU availability depends on the Colab plan and current capacity.

```python
DATA_ADDY = "https://drive.google.com/file/d/1sBc4-CexT3S2cw1LZysrX4BLjVjN6BPt/view?usp=sharing" #@param {type:"string"}
MODEL_ADDY = "https://drive.google.com/file/d/1QLymjt0qnjM2FM-oR5vRYB0B1URcg5wq/view?usp=sharing" #@param {type:"string"}
GIT_ADDY = "https://github.com/a12dongithub/PathOGen" #@param {type:"string"}
GIT_REF = "refactored-structure" #@param {type:"string"}
NUM_IMAGES = 1 #@param {type:"integer"}
STEPS = 20 #@param {type:"integer"}
SEED = 42 #@param {type:"integer"}
```

```python
import os, shutil, subprocess, sys
from pathlib import Path

subprocess.run(["nvidia-smi", "--query-gpu=name,memory.total,driver_version", "--format=csv,noheader"], check=True)
repo = Path("/content/PathOGen")
if repo.exists():
    shutil.rmtree(repo)
subprocess.run(["git", "clone", "--filter=blob:none", GIT_ADDY, str(repo)], check=True)
subprocess.run(["git", "-C", str(repo), "checkout", GIT_REF], check=True)
commit = subprocess.check_output(["git", "-C", str(repo), "rev-parse", "HEAD"], text=True).strip()
print("Running Git commit:", commit)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", str(repo / "experiments" / "requirements-colab.txt")], check=True)
```

```python
command = [
    sys.executable, str(repo / "experiments" / "01_inference_smoke.py"),
    "--data-url", DATA_ADDY,
    "--model-url", MODEL_ADDY,
    "--num-images", str(NUM_IMAGES),
    "--steps", str(STEPS),
    "--seed", str(SEED),
]
subprocess.run(command, check=True)
```
